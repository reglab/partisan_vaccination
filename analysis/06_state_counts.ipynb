```python
import gzip
import pandas as pd
import numpy as np
```

# We have all states included in the AFC generally

```python
with gzip.open('/share/pi/deho-pi/AFC/BQ/GeneratedPatientBaseline_0724.csv.gz', 'rt') as f:
    cols = pd.read_csv(f, nrows=0).columns

print(cols.tolist())
```

```python
with gzip.open('/share/pi/deho-pi/AFC/BQ/GeneratedPatientBaseline_0724.csv.gz', 'rt') as f:
    df = pd.read_csv(f, usecols=['state', 'fips_state'])

print(df.head())
```

AFC contains all 50 states

```python
us_states = {
    "AL","AK","AZ","AR","CA","CO","CT","DE","FL","GA",
    "HI","ID","IL","IN","IA","KS","KY","LA","ME","MD",
    "MA","MI","MN","MS","MO","MT","NE","NV","NH","NJ",
    "NM","NY","NC","ND","OH","OK","OR","PA","RI","SC",
    "SD","TN","TX","UT","VT","VA","WA","WV","WI","WY",
    "DC"
}
df_states = set(df['state'].dropna().unique())
missing_states = sorted(us_states - df_states)
print(missing_states)

unexpected_states = sorted(df_states - us_states)
print(unexpected_states)
```

# We have all states for people with valid household IDs

```python
with gzip.open('/share/pi/deho-pi/AFC/BQ/GeneratedPatientBaseline_0724.csv.gz', 'rt') as f:
    df = pd.read_csv(
        f,
        usecols=['state', 'household_id']
    )

# keep only rows where household_id is NOT NA
df = df[df['household_id'].notna()][['state']]
```

```python
df_states = set(df['state'].dropna().unique())
missing_states = sorted(us_states - df_states)
print(missing_states)

unexpected_states = sorted(df_states - us_states)
print(unexpected_states)
```

## Democratic and Republican household children are complete states. No Republican households are observed in Vermont

```python
df = np.load('/share/pi/deho-pi/AFC/child_party.npy', allow_pickle = True)
```

```python
with gzip.open('/share/pi/deho-pi/AFC/BQ/GeneratedPatientBaseline_0724.csv.gz', 'rt') as f:
    df_state = pd.read_csv(
        f,
        usecols=['state', 'patientuid']
    )
```

```python
df_dem = df[df['party'] == 'Democratic']
df_states = set(df_dem['state'].dropna().unique())
missing_states = sorted(us_states - df_states)
print(missing_states)

unexpected_states = sorted(df_states - us_states)
print(unexpected_states)
```

```python
df_rep = df[df['party'] == 'Republican']
df_states = set(df_rep['state'].dropna().unique())
missing_states = sorted(us_states - df_states)
print(missing_states)

unexpected_states = sorted(df_states - us_states)
print(unexpected_states)
```
